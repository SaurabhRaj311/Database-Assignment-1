#Question 1: Explain the fundamental differences between DDL, DML, and DQL commands in SQL. Provide one example for each type of command.


#**Answer**
**DDL (Data Definition Language)**: Defines, alters, or manages the structural framework of database objects like tables and schemas. It operates on database structure rather than data values.
- Example: CREATE TABLE Students (StudentID INT PRIMARY KEY, Name VARCHAR(50));

**DML (Data Manipulation Language)**: Modifies, adds, or removes actual data records stored within existing database tables.
- Example: INSERT INTO Students (StudentID, Name) VALUES (1, 'Alice');

**DQL (Data Query Language)**: Used specifically to fetch, retrieve, and view data from database tables without altering structures or records.
- Example: SELECT * FROM Students;

#Question 2: What is the purpose of SQL constraints? Name and describe three common types of constraints, providing a simple scenario where each would be useful.

#**Answer**

**Purpose**: SQL constraints enforce rules on columns within a table to maintain data integrity, precision, and reliability across database operations.
- PRIMARY KEY: Uniquely identifies each row in a table and prevents null values.
**Scenario**: Setting CustomerID as a PRIMARY KEY guarantees no two customers share the same ID.
- FOREIGN KEY: Ensures referential integrity between two linked tables by matching values in one table to a PRIMARY KEY in another.
 **Scenario**: Setting CategoryID in a Products table as a FOREIGN KEY linked to Categories ensures products can only be mapped to valid categories.
- UNIQUE: Prevents duplicate values across all rows in a specific column.
**Scenario**: Applying UNIQUE to an Email column prevents multiple users from registering with identical email addresses.

##Question 4: What is a Common Table Expression (CTE) in SQL, and what are its main benefits? Provide a simple SQL example demonstrating its usage.

#**Answers**

A Common Table Expression (CTE) is a temporary, named result set defined using the WITH clause that exists only during the execution of a single SQL statement.

Main Benefits:
- Reusability within a single query.
- Simplifies complex join and subquery logic.
- Improves query readability and maintainability.

```python
import sqlite3
import pandas as pd

# Create in-memory database and test table
conn = sqlite3.connect(':memory:')
cursor = conn.cursor()

cursor.execute('''
CREATE TABLE Orders (
    OrderID INT,
    CustomerID INT,
    TotalAmount DECIMAL(10,2)
)
''')

cursor.executemany('''
INSERT INTO Orders VALUES (?, ?, ?)
''', [
    (1, 101, 150.00),
    (2, 101, 50.00),
    (3, 102, 200.00),
    (4, 103, 120.00),
    (5, 102, 80.00)
])

# Question 4 CTE Query
cte_query = '''
WITH HighValueOrders AS (
    SELECT OrderID, CustomerID, TotalAmount
    FROM Orders
    WHERE TotalAmount > 100
)
SELECT
    CustomerID,
    COUNT(OrderID) AS TotalHighValueOrders
FROM HighValueOrders
GROUP BY CustomerID;
'''

# Execute and display output
df_q4 = pd.read_sql_query(cte_query, conn)
print("--- Question 4 Output ---")
print(df_q4)
```

#Question 5: Describe the concept of SQL Normalization and its primary goals. Briefly explain the first three normal forms (1NF, 2NF, 3NF).

#**Answer**

Concept & Goals: SQL Normalization structures database tables systematically to eliminate redundant data and avoid insertion, update, and deletion anomalies.
- 1NF (First Normal Form): Ensures atomic (indivisible) column values, eliminates repeating groups, and assigns a unique primary key to each row.
- 2NF (Second Normal Form): Meets 1NF requirements and eliminates partial dependencies, meaning all non-key attributes must depend fully on the primary key.
- 3NF (Third Normal Form): Meets 2NF requirements and eliminates transitive dependencies, meaning non-key columns must depend solely on the primary key and not on other non-key columns.

#Question 7: Generate a report showing CustomerName, Email, and the TotalNumberofOrders for each customer. Include customers who have not placed any orders, in which case their TotalNumberofOrders should be 0. Order the results by CustomerName.

```python
#Answer
query = '''
SELECT
    c.CustomerName,
    c.Email,
    COUNT(o.OrderID) AS TotalNumberofOrders
FROM Customers c
LEFT JOIN Orders o ON c.CustomerID = o.CustomerID
GROUP BY c.CustomerID, c.CustomerName, c.Email
ORDER BY c.CustomerName;
'''

df = pd.read_sql_query(query, conn)
print(df)
```

#Question 8: Retrieve Product Information with Category: Write a SQL query to display the ProductName, Price, StockQuantity, and CategoryName for all products. Order the results by CategoryName and then ProductName alphabetically.

```python
#answer
query = '''
SELECT
    p.ProductName,
    p.Price,
    p.StockQuantity,
    c.CategoryName
FROM Products p
JOIN Categories c ON p.CategoryID = c.CategoryID
ORDER BY c.CategoryName ASC, p.ProductName ASC;
'''

df = pd.read_sql_query(query, conn)
print(df)
```

#Question 9: Write a SQL query that uses a Common Table Expression (CTE) and a Window Function (specifically ROW_NUMBER() or RANK()) to display the CategoryName, ProductName, and Price for the top 2 most expensive products in each CategoryName.

```python
#answer
query = '''
WITH RankedProducts AS (
    SELECT
        c.CategoryName,
        p.ProductName,
        p.Price,
        ROW_NUMBER() OVER(PARTITION BY c.CategoryName ORDER BY p.Price DESC) AS Rank
    FROM Products p
    JOIN Categories c ON p.CategoryID = c.CategoryID
)
SELECT
    CategoryName,
    ProductName,
    Price
FROM RankedProducts
WHERE Rank <= 2;
'''

df = pd.read_sql_query(query, conn)
print(df)
```
